```python
import pandas as pd
import ast

# Load the CSV files
matches_df = pd.read_csv('data/epic7_matches.csv')

# Function to parse the pick data from JSON-like strings
def parse_picks(picks_str):
    return sorted(ast.literal_eval(picks_str), key=lambda x: x['pick_order'])

# Apply parsing function to both my_team_picks and enemy_team_picks
matches_df['my_team_picks'] = matches_df['my_team_picks'].apply(parse_picks)
matches_df['enemy_team_picks'] = matches_df['enemy_team_picks'].apply(parse_picks)

# Function to extract hero codes by pick order
def extract_hero_codes(picks):
    return [pick['hero_code'] for pick in picks]

# Extract hero codes for each pick order
matches_df['my_team_hero_codes'] = matches_df['my_team_picks'].apply(extract_hero_codes)
matches_df['enemy_team_hero_codes'] = matches_df['enemy_team_picks'].apply(extract_hero_codes)

# Prepare a new dataframe to hold each draft state and the next hero pick
draft_data = []

# Iterate over each match to create draft states
for _, row in matches_df.iterrows():
    my_picks = row['my_team_hero_codes']
    enemy_picks = row['enemy_team_hero_codes']
    my_bans = ast.literal_eval(row['my_bans'])
    enemy_bans = ast.literal_eval(row['enemy_bans'])
    first_pick_team = row['first_pick_team']
    winner = row['winner']

    # Create draft states for each pick stage (up to 5 picks)
    for i in range(len(my_picks)):
        # Draft state at the i-th pick for my team
        draft_state = {
            'my_picks': my_picks[:i],  # Heroes already picked by my team
            'enemy_picks': enemy_picks[:i],  # Heroes picked by the enemy up to this point
            'my_bans': my_bans,
            'enemy_bans': enemy_bans,
            'first_pick_team': first_pick_team,
            'winner': winner,
            'next_pick': my_picks[i] if i < len(my_picks) else None  # The hero that was picked next
        }
        draft_data.append(draft_state)

# Convert draft data to a DataFrame
draft_df = pd.DataFrame(draft_data)

# Display the first few rows of the transformed draft data
draft_df.head()
```
